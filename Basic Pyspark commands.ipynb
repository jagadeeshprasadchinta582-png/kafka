```python
spark.sql("CREATE CATALOG IF NOT EXISTS my_PYSPARK")
spark.sql("CREATE SCHEMA IF NOT EXISTS my_PYSPARK.my_TEST")
spark.sql("create table if not exists my_PYSPARK.my_TEST.my_table (id int, name string)")
spark.sql("insert into my_PYSPARK.my_TEST.my_table values (1, 'John'), (2, 'Jane') , (3, 'Mary '), (4, 'Bob ')")
```

```python
df = spark.table("my_PYSPARK.my_TEST.my_table")
display(df)
```

```python
df1= spark.table("my_PYSPARK.my_TEST.customers")
display(df1)
```
